```python
import os

local_file = r"C:\Ask My Paper\Data\Copy of AI Engineering-25-125.pdf"

if os.path.exists(local_file):
    print(f"PDF found: {local_file}")
else:
    print("PDF file not found.")
```

Now that the file is loaded in our local folder, we can use Docling to parse
and analyze it.

```python
from docling.document_converter import DocumentConverter,PdfFormatOptions
from docling.datamodal.pipeline_options import PdfPipelineOptions
from docling.datamodal.base_models import InputFormat


pipeline_options = PdfPipelineOptions()
pipeline_options.generate_picture_images = True
res = DocumentConverter.convert(local_file, input_format=InputFormat.PDF, pipeline_options=pipeline_options)
doc = res.document
```
